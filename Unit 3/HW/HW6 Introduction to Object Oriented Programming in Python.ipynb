# ICS4U Practice Question
Name Your File:  `ICS4UcircleObject.py`

Create a Class to represent a Circle with the following information

Attributes:

- Radius (a floating point number)

Constructor:

- Accept the radius as a parameter

Behaviours:

- getArea -> returns the area of the circle

- getCircumference -> returns the circumference of the circle

Write a main program that tests your class.


- Create a few Circle objects with different radius values and then call the behaviour functions to see if they are working correctly.

```python
import math

class Circle: 
    def __init__(self,radius):
        if radius < 0:
            raise ValueError("The Raidus cannot be negative")
        self.radius = float(radius)
    def getArea(self):
        return math.pi * self.radius ** 2 
    def getCircumference(self):
        return 2 * math.pi * self.radius
    def __str__(self):
        return (f"Circle with radius {self.radius}: Area = {self.getArea():.2f}, Circumference = {self.getCircumference():.2f}")
    
def main():
    circle1 = Circle(5)
    circle2 = Circle(10.5)
    circle3 = Circle(2.3)
    print(circle1)
    print(circle2)
    print(circle3)
main()
```

# ICS4U Practice Question
Name Your File:  `ICS4UgeometricCircleObject.py`

Create a Class to represent a Circle with the following information

Attributes:

- Radius (a floating point number)

- CenterX (an int)

- CenterY (an int)

Constructor:


- Accepts 4 integer points on the Cartesian Grid

- Center (x,y) and another point on the circle (x,y)

- Assign a value to the radius based on the inputs, use a helper function

Behaviours:


- getArea -> returns the area of the circle

- getCircumference -> returns the circumference of the circle

- `__str__` -> Display the attributes

Write a main program that tests your class.

- Create a few Circle objects with data points from a file and then call the behaviour functions to see if they are working correctly.

```python
import math

class Circle:
    def __init__(self, centerX, centerY, pointX, pointY):
        self.centerX = int(centerX)
        self.centerY = int(centerY)
        self.pointX = int(pointX)
        self.pointY = int(pointY)
        self.radius = self.calculateRadius()
    
    def calculateRadius(self):
        return math.sqrt((self.pointX - self.centerX)**2 + (self.pointY - self.centerY)**2)
    def getArea(self):
        return math.pi * self.radius ** 2
    def getCircumference(self):
        return 2 * math.pi * self.radius
    def __str__(self):
        return f"Circle with center ({self.centerX}, {self.centerY}) and radius {self.radius:.2f}"

def main():
    try:
        with open('circle_data.txt', 'r') as file:
            for line in file:
                centerX, centerY, pointX, pointY = line.strip().split()
                circle = Circle(centerX, centerY, pointX, pointY)
                print(circle)
                print(f"Area: {circle.getArea():.2f}")
                print(f"Circumference: {circle.getCircumference():.2f}")
                print()
    except FileNotFoundError:
        print("The file 'circle_data.txt' was not found.")

main()
```

# ICS4U Practice Question
Name Your File:  `ICS4UgamePlayer.py`

You are making a game where a  player spins a wheel and rolls a dice to earn points.  Create the player with the following properties:

Attributes:

- Points (total points won during the game)

Constructor:

- No input parameters needed, just set the points to zero

Behaviours:

- spin -> simulates spinning a disk with different colors.  60% of the disk is Blue and earns 3 points, 25% of the disk is Yellow and earns 5 points, 10% of the disk is Red and earns 8 points, 5% of the disk is Black and earns 0 points.  Returns the number of points

- roll -> simulates rolling a 6 sided dice, returns the points earned which is equal to the roll of the dice

Write a main program that tests your class.

- Make 2 players.  Each round of the game consists of a spin and a roll.  Display the winner after 20 rounds.  

```python
import random

class GamePlayer:
    def __init__(self):
        self.points = 0
    
    def spin(self):
        spin_result = random.choices(['Blue', 'Yellow', 'Red', 'Black'], [0.60, 0.25, 0.10, 0.05])[0]
        if spin_result == 'Blue':
            points_earned = 3
        elif spin_result == 'Yellow':
            points_earned = 5
        elif spin_result == 'Red':
            points_earned = 8
        else: 
            points_earned = 0
        self.points += points_earned
        return points_earned

    def roll(self):
        roll_result = random.randint(1, 6)
        self.points += roll_result
        return roll_result

def main():
    player1 = GamePlayer()
    player2 = GamePlayer()

    for round in range(1, 21):
        print(f"Round {round}:")

        # Player 1's turn
        spin1 = player1.spin()
        roll1 = player1.roll()
        print(f"Player 1 spun a wheel and earned {spin1} points, rolled a dice and earned {roll1} points. Total: {player1.points} points")

        # Player 2's turn
        spin2 = player2.spin()
        roll2 = player2.roll()
        print(f"Player 2 spun a wheel and earned {spin2} points, rolled a dice and earned {roll2} points. Total: {player2.points} points")
        
        print()
    if player1.points > player2.points:
        print(f"Player 1 wins with {player1.points} points!")
    elif player2.points > player1.points:
        print(f"Player 2 wins with {player2.points} points!")
    else:
        print(f"It's a tie with both players having {player1.points} points!")

main()
```
